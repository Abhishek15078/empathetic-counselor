### Phase 3 Experiment Analysis

The experiment was conducted using the same user message: "I have been feeling overwhelmed and unable to focus on my work lately." The message was processed three times while changing only the injected emotion context (anxiety, sadness, and neutral). The anxiety-conditioned response focused on fear, worry, and feelings of being overwhelmed, while the sadness-conditioned response emphasized emotional pain, isolation, and discouragement. The neutral response remained more general and focused primarily on the practical impact of feeling overwhelmed without strongly emphasizing a specific emotional state. Since the user message remained constant and only the emotion context changed, the differences in the generated responses demonstrate that emotion-conditioned prompting successfully influences the language model's behavior. This provides evidence that the system can personalize empathetic responses based on detected emotional states, validating the core design objective of the project.
